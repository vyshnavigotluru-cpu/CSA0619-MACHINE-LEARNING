```python
# FIND-S Algorithm

# Training data
# Attributes: Sky, AirTemp, Humidity, Wind, Water, Forecast
# Class: EnjoySport

training_data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

# Number of attributes
num_attributes = len(training_data[0]) - 1

# Initialize the most specific hypothesis
hypothesis = ['0'] * num_attributes

print("Initial Hypothesis:")
print(hypothesis)

# FIND-S algorithm
for row in training_data:

    # Check whether the example is positive
    if row[-1] == 'Yes':

        for i in range(num_attributes):

            # First positive example
            if hypothesis[i] == '0':
                hypothesis[i] = row[i]

            # If attribute values are different, generalize
            elif hypothesis[i] != row[i]:
                hypothesis[i] = '?'

    print("\nAfter processing:", row)
    print("Hypothesis:", hypothesis)

# Final hypothesis
print("\n--------------------------------")
print("FINAL MOST SPECIFIC HYPOTHESIS")
print("--------------------------------")
print(hypothesis)
```
